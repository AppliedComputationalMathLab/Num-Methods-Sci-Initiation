```python
import numpy as np
import struct
import matplotlib.pyplot as plt
import matplotlib as mpl
import numpy as np
```

```python
def f(x: np.float64) -> np.float64:
    return (x**10) + 1 - (x**10)
```

```python
interval = [39.2, 42.4]
x = np.linspace(interval[0], interval[1], 1000)
y = f(x)
```

```python
print(f"value = {(x[y == 2])}")
```

```python
mpl.rcParams['pdf.fonttype'] = 42
mpl.rcParams['ps.fonttype'] = 42

plt.plot(x, y, linewidth=1.3, color='#0077FFFF')
plt.xlim(interval[0], interval[1])

plt.xlabel('x', fontsize=14)
plt.ylabel('f(x) = (x^10 + 1) - x^10', fontsize=14)
plt.yticks(fontsize=12, rotation=0)
plt.xticks(fontsize=12, rotation=0)


centimeters_x = 34
centimeters_y = 11.5

inches_x = centimeters_x / 2.54
inches_y = centimeters_y / 2.54



plt.gcf().set_size_inches(inches_x, inches_y)

plt.grid(True, linewidth=1.0, alpha=1)

plt.savefig(
    "figs/NumError1.pdf",
    format='pdf',
    bbox_inches='tight',
)
```

```python
def getBinary(x: np.float64):
    empacotado = struct.pack('>d', float(x))
    inteiro = int.from_bytes(empacotado, 'big')
    
    binario_puro = f"{inteiro:064b}"
    
    blocos = [binario_puro[i:i+4] for i in range(0, 64, 4)]
    
    return " ".join(blocos)
```

```python
getBinary(42.1**10)
```

```python
getBinary(42.1**10 + 1)
```

```python
getBinary(42.1**10)
```
